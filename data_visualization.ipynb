```python
import numpy as np

months = np.arange(1,5)
organic_traffic = np.array([500,700,800,900])
paid_traffic = np.array([200,300,400,500])
referral_traffic = np.array([100,150,200,240])
```

```python
total_traffic = organic_traffic + paid_traffic + referral_traffic
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10,6))

ax.fill_between(months, 0, organic_traffic, label = "Organic Traffic", color = 'blue', alpha = 0.5)
ax.fill_between(months, organic_traffic + paid_traffic + referral_traffic, label = "paid_traffic", color = 'green', alpha = 0.5)
ax.fill_between(months, organic_traffic + paid_traffic, total_traffic, label = "referral_traffic", color ='red', alpha = 0.6)

ax.set_xlabel('month', fontsize = 12)
ax.set_ylabel('traffic_volume', fontsize = 12)
ax.grid(True, which='both', linestyle='--', linewidth=0.5)
```

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Example dataset (replace with your df)
df = px.data.gapminder()

app = dash.Dash(__name__)

app.layout = html.Div([
    dcc.Dropdown(
        id='dropdown-country',
        options=[{'label': country, 'value': country} for country in df['country'].unique()],
        value='India'
    ),
    dcc.Graph(id='line-plot')
])

@app.callback(
    Output('line-plot', 'figure'),
    Input('dropdown-country', 'value')
)
def update_graph(selected_country):
    filtered_df = df[df['country'] == selected_country]
    fig = px.line(
        filtered_df,
        x='year',
        y='gdpPercap',
        title=f'GDP Per Capita of {selected_country}'
    )
    return fig

if __name__ == '__main__':
    app.run(debug=True)
```
